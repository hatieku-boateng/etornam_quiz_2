# Regression Models with 5000 Metrics

This notebook demonstrates: 
- Simple Linear Regression (SLR)
- Multiple Linear Regression (MLR)
- Logistic Regression (LR)

Each model is trained on a dataset with 5000 samples.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

np.random.seed(42)
n_samples = 5000

x = np.random.uniform(10, 100, n_samples)
noise = np.random.normal(0, 8, n_samples)
y = 25 + 3.5 * x + noise
df_slr = pd.DataFrame({'x': x, 'y': y})
df_slr.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df_slr[['x']], df_slr['y'], test_size=0.2, random_state=42)

slr_model = LinearRegression()
slr_model.fit(X_train, y_train)
y_pred = slr_model.predict(X_test)

slr_metrics = {
    'MAE': mean_absolute_error(y_test, y_pred),
    'MSE': mean_squared_error(y_test, y_pred),
    'RMSE': np.sqrt(mean_squared_error(y_test, y_pred)),
    'R2': r2_score(y_test, y_pred)
}

print('Simple Linear Regression Metrics:')
for metric, value in slr_metrics.items():
    print(f'{metric}: {value:.4f}')

plt.figure(figsize=(8, 5))
plt.scatter(X_test, y_test, color='blue', alpha=0.5, label='Actual')
plt.plot(X_test, y_pred, color='red', linewidth=2, label='Predicted')
plt.title('Simple Linear Regression on 5000 samples')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```

```python
rng = np.random.RandomState(42)
n_samples = 5000
x1 = rng.uniform(0, 100, n_samples)
x2 = rng.uniform(20, 80, n_samples)
noise = rng.normal(0, 5, n_samples)
y_mlr = 15 + 4.2 * x1 - 2.5 * x2 + 0.08 * x1 * x2 + noise

df_mlr = pd.DataFrame({'x1': x1, 'x2': x2, 'y': y_mlr})
X_mlr = df_mlr[['x1', 'x2']]

X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(X_mlr, df_mlr['y'], test_size=0.2, random_state=42)
mlr_model = LinearRegression()
mlr_model.fit(X_train_m, y_train_m)
y_pred_m = mlr_model.predict(X_test_m)

mlr_metrics = {
    'MAE': mean_absolute_error(y_test_m, y_pred_m),
    'MSE': mean_squared_error(y_test_m, y_pred_m),
    'RMSE': np.sqrt(mean_squared_error(y_test_m, y_pred_m)),
    'R2': r2_score(y_test_m, y_pred_m)
}

print('Multiple Linear Regression Metrics:')
for metric, value in mlr_metrics.items():
    print(f'{metric}: {value:.4f}')

print('Model coefficients:', mlr_model.coef_)
print('Intercept:', mlr_model.intercept_)
```

```python
np.random.seed(123)
n_samples = 5000
feature_a = np.random.normal(0, 1, n_samples)
feature_b = np.random.normal(0, 1, n_samples)
logit = -1.2 + 2.0 * feature_a - 1.5 * feature_b
prob = 1 / (1 + np.exp(-logit))
y_bin = (prob > 0.5).astype(int)

df_lr = pd.DataFrame({'feature_a': feature_a, 'feature_b': feature_b, 'target': y_bin})
X_lr = df_lr[['feature_a', 'feature_b']]
y_lr = df_lr['target']

X_train_l, X_test_l, y_train_l, y_test_l = train_test_split(X_lr, y_lr, test_size=0.2, random_state=42, stratify=y_lr)
lr_model = LogisticRegression(max_iter=1000, random_state=42)
lr_model.fit(X_train_l, y_train_l)
y_pred_l = lr_model.predict(X_test_l)
y_prob_l = lr_model.predict_proba(X_test_l)[:, 1]

lr_metrics = {
    'Accuracy': accuracy_score(y_test_l, y_pred_l),
    'Precision': precision_score(y_test_l, y_pred_l),
    'Recall': recall_score(y_test_l, y_pred_l),
    'F1': f1_score(y_test_l, y_pred_l),
    'ROC_AUC': roc_auc_score(y_test_l, y_prob_l)
}

print('Logistic Regression Metrics:')
for metric, value in lr_metrics.items():
    print(f'{metric}: {value:.4f}')

print('Intercept:', lr_model.intercept_)
print('Coefficients:', lr_model.coef_)
```
